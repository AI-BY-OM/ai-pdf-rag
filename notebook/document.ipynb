```python
## directory loader

from pathlib import Path
from langchain_core.documents import Document

documents = []
for file_path in sorted(Path("../data/txt_file").glob("**/*.txt")):
    documents.append(
        Document(
            page_content=file_path.read_text(encoding="utf-8"),
            metadata={"source": str(file_path)},
        )
    )
print(documents)
```

```python
## directory loader
from pathlib import Path
from langchain_core.documents import Document

def load_pdf_documents(folder_path: str):
    documents = []
    for file_path in sorted(Path(folder_path).glob("**/*.pdf")):
        try:
            import fitz
            pdf = fitz.open(str(file_path))
            for page_index, page in enumerate(pdf):
                documents.append(Document(page_content=page.get_text(), metadata={"source": str(file_path), "page": page_index}))
        except Exception:
            from pypdf import PdfReader
            reader = PdfReader(str(file_path))
            for page_index, page in enumerate(reader.pages):
                documents.append(Document(page_content=page.extract_text() or "", metadata={"source": str(file_path), "page": page_index}))
    return documents

pdf_documents = load_pdf_documents("../data/pdf")
print(f"Loaded {len(pdf_documents)} PDF documents")
if pdf_documents:
    print(pdf_documents[0].metadata)
    print(pdf_documents[0].page_content[:200].encode("ascii", errors="ignore").decode("ascii"))
```
